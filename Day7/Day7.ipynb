### Q1. Data Curation: Handling Missing Values

```python
import pandas as pd

data = pd.read_csv("students.csv")

# 1. Detect missing values
print("Missing Values:\n", data.isnull().sum())

# 2. Replace missing Age with mean age
mean_age = data['Age'].mean()
data['Age'] = data['Age'].fillna(mean_age)


# 3. Replace missing Marks with median marks
median_marks = data['Marks'].median()
data['Marks'] = data['Marks'].fillna(median_marks)

print("\nCurated Dataset:\n", data)
```

### Q2. Data Transformation

```python
data['Result'] = data['Marks'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')

print("\nDataset with Result:\n", data)
```

### Q3. Basic Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Histogram of Marks
plt.hist(data['Marks'], bins=5, edgecolor='black')
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# 2. Boxplot of Marks
sns.boxplot(y=data['Marks'])
plt.title("Boxplot of Marks")
plt.show()
```

### Q4. Gender-wise Performance

```python
# Bar chart for average marks by gender
avg_marks = data.groupby("Gender")["Marks"].mean()

avg_marks.plot(kind='bar', color=['blue', 'pink'])
plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")
plt.show()
```

### Q5. Survival Analysis (Titanic Dataset)

```python
# Load Titanic dataset
titanic = sns.load_dataset("titanic")

# 1. Bar chart of survival counts
sns.countplot(x="survived", data=titanic)
plt.title("Survival Counts (0 = Not Survived, 1 = Survived)")
plt.show()

# 2. Survival rate by Gender
sns.barplot(x="sex", y="survived", data=titanic)
plt.title("Survival Rate by Gender")
plt.show()
```

### Q6. COVID-19 Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
covid = pd.read_csv("Covid19-fullPack/covid_19_clean_complete.csv")

# 1. Clean the dataset
covid = covid[['Date', 'Country/Region', 'Confirmed', 'Deaths']]   # select relevant cols
covid = covid.dropna()                                            # remove nulls
covid['Date'] = pd.to_datetime(covid['Date'])                     # ensure datetime

# 2. Select two countries (India & USA as example)
countries = ["India", "US"]

plt.figure(figsize=(12,6))

for country in countries:
    country_data = covid[covid['Country/Region'] == country]
    plt.plot(country_data['Date'], country_data['Confirmed'], label=country)

# Chart formatting
plt.title("COVID-19 Cases Over Time")
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.legend()
plt.grid(True)
plt.show()
```
